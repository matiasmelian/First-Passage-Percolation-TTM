# Bond Percolation

El modelo de *bond percolatio* puede describirse como aquel en el cual los pesos de las aristas tienen una distribución Bernouilli de parámetro $p$. Una arista de peso cero se dice *cerrada* y una de peso uno, *abierta*. La interpretación clásica es que las aristas cerradas han sido ''borradas'', y nos interesa el subgrafo formado por las aristas abiertas, aquellas que han ''sobrevivido''. Una componente conexa de este grafo se denomina *clúster*.

En cada dimensión $d\geq 2$, existe un valor $p_c(d)\in (0,1)$, llamado *parámetro crítico*, tal que si $p<p_c(d)$, entonces casi seguramente no existen clusters infinitos, y si $p > p_c(d)$, entonces c.s. existe un clúster infinito (en cuyo caso decimos que las aristas abiertas *percolan*).

```python
from utils import grid_canvas, fpp, bfs
from random import binomialvariate
from itertools import product

n = 400
m = int(n / 2)

pesos = fpp(n, n, lambda: binomialvariate(1, 0.5))
cluster = bfs(n, n, pesos, (m, m))

bola = [a for a in product(range(0, n + 1), range(0, n + 1)) if cluster[*a]]
grid_canvas(n, n, bola, [], 'teal')
```
